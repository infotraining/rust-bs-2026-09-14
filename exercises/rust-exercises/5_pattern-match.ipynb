# Match

1.

```rust
// Fill the blanks
enum Direction {
    East,
    West,
    North,
    South,
}

fn main() {
    let direction = Direction::South;

    match direction {
        Direction::East => println!("East"),
        Direction::South | Direction::North => { // Matching South or North here
            println!("South or North");
        },
        _ => println!("West"),
    };
}

main();
```

2. Match is an expression, so we can use it in assignments

```rust
fn main() {
    let boolean = true;

    // Fill the blank with a match expression:
    //
    // boolean = true -> binary = 1
    // boolean = false ->  binary = 0
    let binary = boolean as i32;

    let binary = match boolean {
        true => 1,
        false => 0,
    };

    assert_eq!(binary, 1);

    println!("Success!");
}

 main();
```

3. Using match to get the data an enum variant holds.

```rust
// Fill in the blanks
enum Message {
    Quit,
    Move { x: i32, y: i32 },
    Write(String),
    ChangeColor(i32, i32, i32),
}

fn main() {
    let msgs = [
        Message::Quit,
        Message::Move{x:1, y:3},
        Message::ChangeColor(255,255,0)
    ];

    for msg in msgs {
        show_message(msg)
    }

    println!("Success!");
} 

fn show_message(msg: Message) {
    match msg {
        Message::Move{ x, y } => { // match  Message::Move
            assert_eq!(x, 1);
            assert_eq!(y, 3);
            println!("Move message with x: {}, y: {}", x, y);
        },
        Message::ChangeColor(_, g, b) => {
            assert_eq!(g, 255);
            assert_eq!(b, 0);
            println!("ChangeColor message with g: {}, b: {}", g, b);
        },
        Message::Write(s) => { println!("{}", s); },
        Message::Quit => println!("no data in these variants")
    }
}

main();
```

```rust
struct Data {
    items: Vec<i32>,
    current_pos: usize,
}

impl Data {
    fn new(items: Vec<i32>) -> Self {
        Data {
            items,
            current_pos: 0,
        }
    }

    fn next(&mut self) -> Option<i32> {
        if self.current_pos < self.items.len() {
            let result = self.items[self.current_pos];
            self.current_pos += 1;
            return Some(result);
        }
        else {
            return None;
        }
    }
}
```

```rust
let mut data = Data::new(vec![1, 2, 3]);
```

```rust
while let Some(value) = data.next() {
    println!("{}", value);
}
```
