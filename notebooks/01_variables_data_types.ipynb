# Variables and Data Types

## Declaring variables

* Variables in Rust are declared using the `let` keyword.
* Rust is a statically typed language, which means that you need to specify the type of the variable when you declare it: 

```rust
let age: u8 = 30;
let pi: f64 = 3.14159;
let is_ok: bool = true;
```

* Or the compiler can infer the type of the variable from the value assigned to it:

```rust
let n = 0; // n: i32
let e = 2.718f64; // e: f64
let is_done = false; // is_done: bool
```

* Variables are **immutable** by default

```rust
let step = 1;
println!("step={}", step);
step += 2
```

* All variables must be initialized before use

```rust
let id: u32;
println!("id={}", id);
```

* but the initialization can be done later

```rust
let number: i32;
println!("number={}", number);

let programming_language;
programming_language = "Rust";
println!("{} is programming language", programming_language);
```

## Mutable variables

* You can make a variable mutable by using the `mut` keyword

```rust
let mut count = 0;
println!("Initial count: {}", count);
count += 1;
println!("Updated count: {count}");
count += 1;
println!("Updated count: {count}");
```

## Shadowing

* You can declare a new variable with the same name as an existing variable, and the new variable will "shadow" the existing variable

```rust
fn main() { 
    let outer = 42; 
     { // start of code block 
          let inner = 3.14; 
          println!("block variable: {}", inner); 
          let outer = 99; // shadows the first outer variable 
          println!("block variable outer: {}", outer); 
     } // end of code block 
     println!("outer variable: {}", outer); 
  } 

  main();
```

## Constants

* Constants are:
  * always immutable
  * evaluated at compile time.
* Constants can be declared in any scope, including the global scope
* By convention, constants are named using uppercase letters with underscores separating words (e.g., `MAX_SIZE`).

```rust
const SECONDS_IN_MINUTE: u32 = 60;
const MINUTES_IN_HOUR: u32 = 60;
const SECONDS_IN_HOUR: u32 = SECONDS_IN_MINUTE * MINUTES_IN_HOUR;
```

```rust
println!("seconds in one hour: {}", SECONDS_IN_HOUR);
```

## Static Variables

* Rust also has static variables, which are similar to constants but have some differences:
  * Lifetimes: static variables have a `'static` lifetime, meaning they live for the entire duration of the program.
  * Mutability: static variables can be mutable if declared with static mut, but this requires unsafe code to access or modify them.


```rust
static MESSAGE: &str = "Hello Rust!";
static mut COUNTER: u32 = 0;

fn click() {
    unsafe {
        COUNTER += 1;
    }
}

fn main() {
    println!("{}", MESSAGE);
    click();
    click();
    unsafe {
        println!("COUNTER: {}", *std::ptr::addr_of!(COUNTER));
    }
}

main();
```

* Rust discourages the use of mutable static variables due to potential data races and undefined behavior.
* If you need a global mutable state, consider using `std::sync::Mutex` or `std::sync::Atomic` types instead.

### Overflow

* In debug mode, Rust includes checks for integer overflow. If overflow occurs, Rust will panic and exit the program. 

```rust
let big_number = std::i32::MAX;
big_number
```

```rust
let next_number = big_number + 1; // panic: arithmetic operation overflowed
```

* In release mode overflow will wrap to negative value (unlike in C++ where it is undefined behavior)

#### Controlling Overflow Behavior

* To control overflow behavior explicitly, you can use methods:
  * `checked_add`, `checked_sub`, `checked_mul` etc. for checked arithmetic
  * `wrapping_add`, `wrapping_sub`, `wrapping_mul` etc. for wrapping arithmetic
  * `saturating_add`, `saturating_sub`, `saturating_mul` etc. for saturating arithmetic

```rust
let safe_next_number: Option<i32> = big_number.checked_add(1);
safe_next_number
```

```rust
let next_number = big_number.wrapping_add(1); // -2147483648
next_number
```

```rust
let next_number = big_number.saturating_add(1);
next_number
```

## The Character Type

* `char` char in Rust can represent any Unicode scalar value, including alphabetic characters, symbols, control characters, and more.
* `char` literals are enclosed in single quotes ('A').


```rust
fn main() {
    let letter = 'A';  // A single character
    let emoji = '😊';  // An emoji character
    let special_character = 'ñ';  // A special alphabetic character

    println!("Letter: {}", letter);  // Output: Letter: A
    println!("Emoji: {}", emoji);    // Output: Emoji: 😊
    println!("Special Character: {}", special_character);  // Output: Special Character: ñ
}

main();
```

* `char` is always four bytes in size

```rust
std::mem::size_of::<char>()
```

```rust
let v = vec!['h', 'e', 'l', 'l', '😊'];

// five elements times four bytes for each element
assert_eq!(20, v.len() * std::mem::size_of::<char>());
```

## Tuples

* Tuples are a way of grouping together a number of values with a variety of types into one compound type.
* Tuples have a fixed length: once declared, they cannot grow or shrink in size.

```rust
let person: (&str, &str, i32) = ("James", "Bond", 42);
```

* Tuples are indexed by position, starting at zero

```rust
person.0
```

```rust
person.1
```

```rust
person.2
```

* Destructuring a tuple

```rust
let person = ("James", "Bond", 42);
let (first, last, age) = person;
println!("first={}, last={last}, age={age}", first);

let (first_name, last_name, ..) = person;
println!("first_name={}, last_name={}", first_name, last_name);
```

* Swap for tuples

```rust
let mut a = 1;
let mut b = 2;
println!("a={}, b={}", a, b);

(a, b) = (b, a);
println!("a={}, b={}", a, b);
```

## Extra - Utilities for working with variables and data types

* Calculating size of variables and data types
  

```rust
let x = 42; // Example variable
let size_of_x = std::mem::size_of_val(&x);
println!("Size of x: {} bytes", size_of_x);

let y = String::from("text");
let size_of_y = std::mem::size_of_val(&y);
println!("Size of y: {} bytes", size_of_y);

println!("Size of String: {} bytes", std::mem::size_of::<String>());
println!("Size of &str: {} bytes", std::mem::size_of::<&str>());
```

* `TypeId`
  * `TypeId` is used to get a unique identifier for a type at runtime. It can be used to compare if two types are the same.

```rust
use std::any::TypeId;

let id = TypeId::of::<i32>();
println!("{id:?}");
```

* `TypeId` can be used to assert if two types are the same or different at runtime using `assert_eq!` and `assert_ne!`.

```rust
assert_eq!(TypeId::of::<i32>(), TypeId::of::<i32>());
assert_ne!(TypeId::of::<i32>(), TypeId::of::<u32>());
```

* How to get the `TypeId` of a type at runtime using `TypeId::of::<T>()`? We can use the `type_id` method on a reference to the value cast as `&dyn Any`.

```rust
use std::any::Any;

let a = 42;

(&a as &dyn Any).type_id() // gets the TypeId of the value `a` at runtime
```

```rust
let a = 42;
let b = String::from("Hello");

assert_eq!((&a as &dyn Any).type_id(), TypeId::of::<i32>());
assert!((&a as &dyn Any).is::<i32>()); // simplified type check

assert_eq!((&b as &dyn Any).type_id(), TypeId::of::<String>());
assert!((&b as &dyn Any).is::<String>());
```

# Extra - Formatting in Rust

Rust provides a powerful and flexible formatting system based on **format strings** and **placeholders** inside `{}`.  
The three most common macros are:

- **`println!`** — prints formatted text to standard output  
- **`format!`** — returns a formatted `String`  
- **`write!` / `writeln!`** — writes formatted text into any buffer implementing `std::fmt::Write`

Below are the essential examples and formatting options.

## Basic formatting



```rust
println!("Hello, {}!", "world");
let s: String = format!("Sum = {}", 2 + 3);
println!("{}", s);
```

## Debug formatting

```rust
let v = vec![1, 2, 3];

println!("{:?}", v);   // debug
println!("{:#?}", v);  // pretty debug (multiline)
```

## Positional and named arguments

```rust
println!("{1} {0}", "a", "b");               // "b a"
println!("{name} is {age}", name="Alice", age=30);
```

## Alignment and width

```rust
println!("{:>8}", 42);       // right-align, width 8
println!("{:<8}", 42);       // left-align
println!("{:^8}", 42);       // center
println!("{:*^8}", 42);      // center with '*' fill
```

## Numeric formatting

```rust
println!("{:04}", 7);        // zero‑pad → "0007"
println!("{:+}", 5);         // always show sign → "+5"
println!("{:#x}", 255);      // hex with prefix → "0xff"
println!("{:#b}", 5);        // binary with prefix → "0b101"
```

## Precision


```rust
println!("{:.2}", 3.14159);  // "3.14"
println!("{:.5}", "hello world"); // "hello"
```

## Escaping braces

```rust
println!("{{}}");  // prints "{}"
```

## Using `write!` and `writeln!`

```rust
use std::fmt::Write;

let mut out = String::new();

write!(out, "Hello {}!", "Rust").unwrap();
writeln!(out, " Number: {}", 42).unwrap();

println!("{}", out);
```
